***STUDENT PERFORMANCE ANALYSIS***

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("/content/StudentsPerformance.csv",encoding='latin1')
```

*BASIC STATISTICS*

```python
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
df=df.dropna()
```

```python
df=df.drop_duplicates()
```

```python
df.isnull().sum()
```

```python
avg_math=df["math score"].mean()
```

```python
avg_math
```

```python
avg_read=df["reading score"].mean()
avg_read
```

```python
avg_write=df["writing score"].mean()
avg_write
```

```python
averages={"Math Average":avg_math,"Reading Average":avg_read,"Writing Average":avg_write}
```

```python
averages
```

```python
df["total_score"]=df["math score"]+df["reading score"]+df["writing score"]
df["total_score"]
```

```python
df["total_score"].head()
```

```python
top_students=df.sort_values(by="total_score",ascending=False).head(5)
top_students
```

*VISUALIZATIONS*

```python
df.groupby("race/ethnicity")["reading score"].sum().plot(kind='bar')
```
