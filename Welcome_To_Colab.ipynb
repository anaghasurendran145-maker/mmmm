*VISUALIZATIONS*

```
Highest_10 = df.sort_values(by='Confirmed', ascending=False).head(10)
```

```
Highest_10
```

```
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 8))
```

```
sns.barplot(x='Confirmed', y='Country/Region', data=Highest_10,
            label='Recovered', color='g', edgecolor='black')

plt.title('Top 10 Countries: Confirmed Cases (COVID-19)', fontsize=16)
plt.xlabel('Number of Cases', fontsize=12)
plt.ylabel('Country', fontsize=12)
plt.legend(loc='lower right')
plt.tight_layout()
plt.show()
```

```
top_recovery_rates =df.sort_values(by='Recovered / 100 Cases', ascending=False).head(20)
```

```
top_recovery_rates
```

```
sns.barplot(
    x='Recovered / 100 Cases',
    y='Country/Region',
    data=top_recovery_rates,
    palette='viridis'
)
plt.title('Top 20 Countries by Recovery Rate (Min 10,000 Confirmed Cases)', fontsize=16)
plt.xlabel('Recovery Rate (%)', fontsize=12)
plt.ylabel('Country', fontsize=12)
plt.tight_layout()
plt.show()
```

```
top_death_rates =df.sort_values(by='Deaths / 100 Cases', ascending=False).head(20)
```

```
top_death_rates
```

```
sns.barplot(
    x='Deaths / 100 Cases',
    y='Country/Region',
    data=top_death_rates,
    palette='viridis'
)
plt.title('Top 20 Countries by Death Rate (Min 10,000 Confirmed Cases)', fontsize=16)
plt.xlabel('Death Rate (%)', fontsize=12)
plt.ylabel('Country', fontsize=12)
plt.tight_layout()
plt.show()
```

```
high_new_cases=df.sort_values(by='New cases', ascending=False).head(20)
```

```
high_new_cases
```

```
sns.barplot(
    x='New cases',
    y='Country/Region',
    data=high_new_cases,
    palette='viridis'
)
plt.title('Top 20 Countries by New Cases (Min 10,000 Confirmed Cases)', fontsize=16)
plt.xlabel('New Cases', fontsize=12)
plt.ylabel('Country', fontsize=12)
plt.tight_layout()
plt.show()
```

```
active_cases =df.sort_values(by='Active', ascending=False).head(20)
```

```
active_cases
```

```
sns.barplot(
    x='Active',
    y='Country/Region',
    data=active_cases,
    palette='viridis'
)
plt.title('Top 20 Countries by Active Cases (Min 10,000 Confirmed Cases)', fontsize=16)
plt.xlabel('New Cases', fontsize=12)
plt.ylabel('Country', fontsize=12)
plt.tight_layout()
plt.show()
```

*INSIGHTS*

High Fatality: Yemen stands out with a significantly higher death rate (nearly 25-30%) compared to others.

Volume Drivers: The US, India, and Brazil lead in Confirmed Cases, New Cases, and Active Cases.

The Recovery Rate chart shows a very different set of leaders.
Countries like Dominica, Grenada, and Iceland show recovery rates near 100%.

The New Cases and Active Cases graphs tell us that:

​The US is the primary outlier in almost every volume-based category, maintaining the highest number of active and new infections.

​India and Brazil consistently hold the second and third spots, showing that the pandemic was most aggressive in these large, populous regions during this data snapshot.

​In the Active Cases chart, there is a massive drop-off after the top three (US, Brazil, India), indicating a high concentration of the global active caseload in just a few nations.
