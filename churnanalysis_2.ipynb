```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df = pd.read_csv("/content/churn.csv", encoding='latin1')
df = df.dropna()
df = df.drop_duplicates()

le = LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])
le.classes_
df['Geography'] = le.fit_transform(df['Geography'])
le.classes_
```

**CHURN RELATIONSHIP ANALYSIS**

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Exited', y=df['Age'], data=df, palette='Set1')
plt.title(f'Relationship: Age vs Churn Status', fontsize=15)
plt.xlabel('Churn Status (0 = Stayed, 1 = Exited)', fontsize=12)
plt.ylabel('Age', fontsize=12)
plt.xticks([0, 1], ['Stayed', 'Exited'])
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Exited', y=df['CreditScore'], data=df, palette='Set1')
plt.title(f'Relationship: CreditScore vs Churn Status', fontsize=15)
plt.xlabel('Churn Status (0 = Stayed, 1 = Exited)', fontsize=12)
plt.ylabel('CreditScore', fontsize=12)
plt.xticks([0, 1], ['Stayed', 'Exited'])
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Exited', y=df['Balance'], data=df, palette='Set1')
plt.title(f'Relationship: Balance vs Churn Status', fontsize=15)
plt.xlabel('Churn Status (0 = Stayed, 1 = Exited)', fontsize=12)
plt.ylabel('Balance', fontsize=12)
plt.xticks([0, 1], ['Stayed', 'Exited'])
plt.show()
```

**INSIGHTS**

```python
numeric_df = df.drop(['RowNumber', 'CustomerId', 'Surname'], axis=1)
corr = numeric_df.corr()
plt.figure(figsize=(10,8))
sns.heatmap(corr,annot=True,cmap='coolwarm')
```

1.**AGE AND CHURN**

. The distribution of age of people who are leaving lies between 40-60.

.Younger customers are significantly more loyal, whereas middle-aged to older customers show a much higher probability of leaving.

2.**ACCOUNT BALANCE AND CHURN**

.The people who are staying are with 0balance account.

.Those who "Exited," the distribution is a near-perfect bell curve centered around $100,000 to $125,000.

3.**CREDIT SCORE AND CHURN**

.It is a neutral factor,The distribution of credit scores for both those who stayed and those who left is almost identical, centered around 650.

4.**TENURE AND CHURN**

.Whether the customer is a long-term customer or short-customer, they are almost equally-likely to exit.

5.**ESTIMATED SALARY AND CHURN**

.As in the same case of tenure and churn,High earners and low earners are exiting at roughly the same rates.

**SUMMARY**

.A middle-aged customer (45-60) with a high account balance (that is, account that is not 0balance), regardless of how long they have been with the bank or how much they earn are exiting the bank more often.
