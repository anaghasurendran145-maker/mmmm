*SALES DATA ANALYSIS*



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales_data = pd.read_csv("/content/sales_data_sample.csv",encoding='latin1')
```

**BASIC** **STATISTICS**

```python
sum=sales_data['SALES'].sum()
sum
```

```python
mean=sales_data.SALES.mean()
mean
```

```python
max=sales_data.SALES.max()
```

```python
max
```

```python
min=sales_data.SALES.min()
min
```

***SALES BY REGION***

```python
plt.figure(figsize=(10,6))
sns.barplot(data=sales_data,x='TERRITORY',y='SALES',color='skyblue',edgecolor='black')
plt.title("Total Sales by Region",fontsize=14)
plt.xlabel("Territory",fontsize=12)
plt.ylabel("Total Sales",fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show
```

Here from the graph it is explicitly showing that, the top-performing region is undoubtedly JAPAN!!!
