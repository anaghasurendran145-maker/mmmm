*SALES DATA ANALYSIS*



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales_data = pd.read_csv("/content/sales_data_sample.csv",encoding='latin1')
```

**BASIC** **STATISTICS**

```python
sales_data.head()
```

```python
sales_data.info()
```

```python
sales_data.describe()
```

```python
sales_data=sales_data.dropna()
```

```python
sales_data=sales_data.drop_duplicates()
```

```python
sales_data["ORDERDATE"]=pd.to_datetime(sales_data["ORDERDATE"])
```

```python
sum=sales_data['SALES'].sum()
sum
```

```python
mean=sales_data.SALES.mean()
mean
```

```python
max=sales_data.SALES.max()
```

```python
max
```

```python
min=sales_data.SALES.min()
min
```

```python
sales_data.groupby("PRODUCTLINE")["SALES"].sum().sort_values(ascending=True)
```

```python
sales_data.groupby("COUNTRY")["SALES"].sum().sort_values(ascending=True)
```

***SALES BY REGION***

```python
plt.figure(figsize=(10,6))
sns.barplot(data=sales_data,x='COUNTRY',y='SALES',color='skyblue',edgecolor='black')
plt.title("Total Sales by Region",fontsize=14)
plt.xlabel("Country",fontsize=12)
plt.ylabel("Total Sales",fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show
```

Here from the graph it is explicitly showing that, the top-performing is country undoubtedly USA!!!

```python
plt.figure(figsize=(10,6))
sns.barplot(data=sales_data,x='PRODUCTLINE',y='SALES',color='skyblue',edgecolor='black')
plt.title("Total Sales by Category",fontsize=14)
plt.xlabel("Category",fontsize=12)
plt.ylabel("Total Sales",fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show
```

Here from the graph it is explicitly showing that, the top-performing is category is undoubtedly Classic cars!!!

```python
sales_data.groupby("PRODUCTLINE")["SALES"].sum().plot(kind='pie')
```

```python
sales_data.groupby("TERRITORY")["SALES"].sum().plot(kind='pie')
```

Here from the graph it is explicitly showing that, the top-performing territory is undoubtedly EMEA!!!
